```python
import numpy as np
from scipy.special import expit
from scipy.optimize import minimize

import matplotlib as mpl
import matplotlib.pyplot as plt
```

Recall our logistic cost function:
$$
J(\theta) = -\frac{1}{m}\left[\sum_{i=1}^m y^{(i)}\log h_\theta(x^{(i)}) + (1 - y^{(i)})\log (1 - h_\theta(x^{(i)})\right]
$$

When using high-order polynomials in our cost function, this can tend to lead to over-fitting, where $\theta_i$'s get large to create tightly turning decision boundaries.  We can avoid this through *regularization*, i.e., by adding a penalty term for large $\theta_i$'s.

$$
J(\theta) = -\frac{1}{m}\left[\sum_{i=1}^m y^{(i)}\log h_\theta(x^{(i)}) + (1 - y^{(i)})\log (1 - h_\theta(x^{(i)})\right] + \frac{\lambda}{2m}\sum_{j=1}^n \theta_j^2
$$

or vectorized

$$
J(\theta) = \frac{1}{m}\left(y^T\log(g(X\theta)) + (1-y)^T\log(1 - g(X\theta)\right) + \frac{\lambda}{2m}\sum_{j=1}^n\theta_j^2
$$

```python
def cost(theta, X, y, reg):
    m = y.size
    h = expit(X.dot(theta))

    J = -1/m * (y.T.dot(np.log(h)) + (1 - y).T.dot(np.log(1 - h)) + reg/(2*m) * np.sum(np.square(theta)))
    J = J[0]

    if np.isnan(J):
        return np.inf
    return J

def cost_gradient(theta, X, y, reg):
    m = y.size
    theta = theta.reshape(-1, 1)
    h = expit(X.dot(theta))

    gradient = 1/m * X.T.dot(h - y) + reg/m * np.concatenate(([[0]], theta[1:])).reshape(-1, 1)
    return gradient.flatten()
```

## One-vs-All

Let's build a classifier capable of identifying each of the 10 digits.  To do this we'll use the one-vs-all technique, where we will <font color='orange'>train 10 separate classifies, each tasked with identifying one particular digit from the rest, then maximize over the returned probabilities to predict a particular digit provided an image.

```python
training = np.random.choice([True, False], size=len(digits.data), p=[0.9, 0.1])
validating = ~training
```

```python
# count non-zero values
m = np.count_nonzero(training)
mtest = np.count_nonzero(validating) 
# stack a column of 1's the size of the non-zeri values in training with the training data
X = np.column_stack([np.ones((m, 1)), digits.data[training]])
# do the same with the validating data
Xtest = np.column_stack([np.ones((mtest, 1)), digits.data[validating]])

# make these into 1D arrays
y = digits.target[training].reshape(-1, 1)
ytest = digits.target[validating].reshape(-1, 1)
```

### <font color='orange'> Define our model
    
<font color='orange'>train 10 separate classifies, each tasked with identifying one particular digit from the rest, then maximize over the returned probabilities to predict a particular digit provided an image. 10 different logistic methods. Is it in or out ten times simultaneously.

```python
def one_v_all(X, y, n_classes, reg=0.1):
    n_features = X.shape[1] # number of features is the number of columns of X

    # theta_0 is an array of zeroes the saize of n_features
    theta0 = np.zeros(n_features)  # I'd like this to be shape (n_features, 1), but scipy complains now
    # thetas is an array of zeroes n_classes (given) x n_features
    thetas = np.zeros((n_classes, n_features))

    # # for each class in n_classes (given) minimize the cost ...evaluated at theta0?
    # for cls in np.arange(n_classes):
    #     res = minimize(cost, theta0, args=(X, y==cls, reg), jac=cost_gradient, options={'maxiter':50})
    #     thetas[cls] = res.x
    # return thetas
    # Loop over each class for one-vs-rest logistic regression
    for cls in np.arange(n_classes):

        # -----------------------------------------------------------
        # Train a binary classifier: "is y == cls?"
        #
        # scipy.minimize optimizes the logistic regression cost
        # function for this particular class by adjusting theta.
        #
        # cost(...)         → logistic loss + regularization
        # theta0            → initial parameter guess (all zeros)
        # args              → arguments passed to cost():
        #                       X        = feature matrix
        #                       y == cls = boolean labels (one-vs-rest)
        #                       reg      = regularization strength
        # jac               → gradient function for fast convergence
        # maxiter=50        → stop after 50 iterations
        # -----------------------------------------------------------
        res = minimize(
            cost,              # loss function
            theta0,            # initial θ vector
            args=(X, y == cls, reg),   # extra args to cost()
            jac=cost_gradient,         # derivative of cost
            options={'maxiter': 50}    # optimization settings
        )

        # -----------------------------------------------------------
        # Store optimized θ for this class.
        #
        # thetas is a matrix where:
        #   thetas[0] → parameters for class 0 vs rest
        #   thetas[1] → parameters for class 1 vs rest
        #   ...
        # -----------------------------------------------------------
        thetas[cls] = res.x

    # Return all class parameter vectors
    return thetas
```

```python
with np.errstate(divide='ignore'):
    theta = one_v_all(X, y, 10)
```

```python
theta.shape
```

```python
def predict_one_v_all(thetas, X):
    Ps = expit(X.dot(theta.T))
    
    return np.argmax(Ps, axis=1)
```

```python
pred = predict_one_v_all(theta, X)
predtest = predict_one_v_all(theta, Xtest)

training_acc = np.mean(pred == y.ravel())*100
validation_acc = np.mean(predtest == ytest.ravel())*100
print(training_acc, validation_acc)
```

```python
i = np.random.choice(Xtest.shape[0])

plt.imshow(Xtest[i, 1:].reshape(-1, 8), cmap='Greys')
print("prediction: {}".format(predict_one_v_all(theta, Xtest[i:i+1])))
print("truth: {}".format(ytest[i]))
```
