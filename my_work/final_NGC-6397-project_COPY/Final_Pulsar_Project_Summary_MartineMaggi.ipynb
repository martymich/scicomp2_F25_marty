## Prepare the Training Data

I use the five paramaters highlighted earlier:
    - Period (logged), logP0
    - Period Derivative, logP1
    - Characteristic Age, AGE
    - Spin Down Energy Loss Rate, EDOT
    - Magnetic Field Strength at Surface, BSURF
    
I classify MSP as 1 and Not-MSP as 0.

Then the data is balanced due to the asymmetrical size of the classes. There are way more CPs than MSPs.
